```python
import pandas as pd
import sqlite3

# This is the path to the file Molemi told you to use
file_path = '/content/drive/MyDrive/WTM-Data-Science-Group-B/data/processed/cleaned_retail_data.csv'

# Load the data
df = pd.read_csv(file_path)

# Create the SQL database
conn = sqlite3.connect('retail.db')
df.to_sql('sales', conn, if_exists='replace', index=False)

print("Database Ready! You can proceed.")
```

These queries calculate the top sellers, monthly revenue trends, top countries, slow-moving products, and top revenue products. Note: The drop in December 2011 is because the data ends on 9 December 2011.

```python
# QUERY 1: TOP 10 PRODUCTS
query1 = """
SELECT StockCode, MAX(Description) as Description, SUM(Quantity) as Total_Sold
FROM sales
GROUP BY StockCode
ORDER BY Total_Sold DESC
LIMIT 10;
"""
df1 = pd.read_sql_query(query1, conn)
df1.to_csv('/content/drive/MyDrive/WTM-Data-Science-Group-B/outputs/top_10_sql_check.csv', index=False)

# QUERY 2: MONTHLY SALES TREND
query2 = """
SELECT substr(InvoiceDate, 1, 7) as Month, SUM(Quantity * UnitPrice) as Total_Revenue
FROM sales
GROUP BY Month
ORDER BY Month;
"""
df2 = pd.read_sql_query(query2, conn)
df2.to_csv('/content/drive/MyDrive/WTM-Data-Science-Group-B/outputs/monthly_sales_trend.csv', index=False)

# QUERY 3: SALES BY COUNTRY
query3 = """
SELECT Country, SUM(Quantity * UnitPrice) as Total_Revenue
FROM sales
GROUP BY Country
ORDER BY Total_Revenue DESC
LIMIT 5;
"""
df3 = pd.read_sql_query(query3, conn)
df3.to_csv('/content/drive/MyDrive/WTM-Data-Science-Group-B/outputs/sales_by_country.csv', index=False)

# QUERY 4: SLOW-MOVING PRODUCTS
query4 = """
SELECT StockCode, MAX(Description) as Description, SUM(Quantity) as Total_Sold
FROM sales
GROUP BY StockCode
ORDER BY Total_Sold ASC
LIMIT 10;
"""
df4 = pd.read_sql_query(query4, conn)
df4.to_csv('/content/drive/MyDrive/WTM-Data-Science-Group-B/outputs/slow_moving_products.csv', index=False)

# QUERY 5: TOP REVENUE PRODUCTS
query5 = """
SELECT StockCode, MAX(Description) as Description, SUM(Quantity * UnitPrice) as Total_Revenue
FROM sales
GROUP BY StockCode
ORDER BY Total_Revenue DESC
LIMIT 10;
"""
df5 = pd.read_sql_query(query5, conn)
df5.to_csv('/content/drive/MyDrive/WTM-Data-Science-Group-B/outputs/top_revenue_products.csv', index=False)

print("Done! All files saved to Drive with the updated data.")
```
